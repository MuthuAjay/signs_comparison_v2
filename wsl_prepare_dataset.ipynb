```python
from pathlib import Path
import cv2
import numpy as np


# list the images which has more than 80% of pixels as white

def list_images_with_high_white_pixels(image_dir, threshold=0.8):
    """
    List images with more than 80% white pixels.
    
    Args:
        image_dir: Directory containing images to check.
        threshold: Percentage of white pixels above which the image will be listed.
        
    Returns:
        List of image paths that exceed the white pixel threshold.
    """
    image_dir = Path(image_dir)
    if not image_dir.exists() or not image_dir.is_dir():
        raise ValueError(f"Invalid directory: {image_dir}")
    
    high_white_images = []
    
    for image_path in image_dir.glob('*'):
        if image_path.suffix.lower() not in ['.jpg', '.jpeg', '.png', '.bmp', '.tiff']:
            continue  # Skip non-image files
        
        # Load the image
        image = cv2.imread(str(image_path), cv2.IMREAD_GRAYSCALE)
        if image is None:
            print(f"Could not load image: {image_path}")
            continue
        
        # Calculate percentage of white pixels
        total_pixels = image.size
        white_pixels = np.sum(image >= 240)  # Assuming white is near 255
        white_percentage = white_pixels / total_pixels
        
        if white_percentage > threshold:
            high_white_images.append(image_path)
            print(f"{image_path} has {white_percentage:.2%} white pixels")
    
    return high_white_images

list_of_images_75_thres = list_images_with_high_white_pixels(
    "/home/eyhyd/signature_comparison_project/signver/datasets/cropped_masks",
    threshold=0.75
)

```

```python
# remove the files from the images and masks directories
def remove_files_from_directories(file_paths, image_dir, mask_dir):
    """
    Remove files from both image and mask directories.
    
    Args:
        file_paths: List of file paths to remove (without directory).
        image_dir: Directory containing images.
        mask_dir: Directory containing masks.
    """
    
    image_dir = Path(image_dir)
    mask_dir = Path(mask_dir)
    
    for file_name in file_paths:
        # replace _mask with ''
        image_file_name = file_name.replace('_mask', '')
        # Construct full paths
        image_path = image_dir / image_file_name
        mask_path = mask_dir / file_name
        
        if image_path.exists():
            image_path.unlink()  # Remove the image file
            print(f"Removed image: {image_path}")
        
        # check if mask_path exists
        if mask_path.exists():
            mask_path.unlink()
            print(f"Removed mask: {mask_path}")

            
```

```python
# create a new directory and move the images to that directory
def move_images_to_directory(image_paths, target_dir):
    """
    Move images to a specified directory.
    
    Args:
        image_paths: List of image paths to move.
        target_dir: Directory to move the images to.
    """
    target_dir = Path(target_dir)
    target_dir.mkdir(parents=True, exist_ok=True)
    
    for image_path in image_paths:
        if not image_path.exists():
            print(f"Image does not exist: {image_path}")
            continue
        
        new_path = target_dir / image_path.name
        image_path.rename(new_path)  # Move the file
        print(f"Moved {image_path} to {new_path}")

move_images_to_directory(list_of_images_75_thres, "/home/eyhyd/signature_comparison_project/signver/datasets/high_white_images")

# remove the files from the images and masks directories

remove_files_from_directories(
    [p.name for p in list_of_images_60_thres],
    "/home/eyhyd/signature_comparison_project/signver/datasets/cropped_images",
    "/home/eyhyd/signature_comparison_project/signver/datasets/cropped_masks"
)



```
