```python
import pandas as pd
import numpy as np
import os

data_path = r"C:\Users\EYLAB\Downloads\signver\full_data.csv"

mapping = r"C:\Users\EYLAB\Downloads\signver\updated_image_ids.csv"

df = pd.read_csv(data_path)

mapping_df = pd.read_csv(mapping)

df.head()
```

```python
folder_path = r"C:\Users\EYLAB\Downloads\signver\images"

id_to_img_dict = {row['id']: row['file_name'] for _, row in mapping_df.iterrows()}
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
import ast  # safer than eval

def get_image_path(id):
    image_name = id_to_img_dict[id]
    image_path = os.path.join(folder_path, image_name)
    return image_path

def get_bbox(id):
    bbox = df[df["image_id"] == id]['bbox']
    return bbox

def show_image_with_bbox(id):
    image_path = get_image_path(id)
    bboxs = get_bbox(id)

    # Read the image
    image = cv2.imread(image_path)
    if image is None:
        raise ValueError(f"Image not found: {image_path}")
    
    height, width = image.shape[:2]
    
    for bbox in bboxs:
        if isinstance(bbox, str):
            # Use literal_eval for safety
            bbox = ast.literal_eval(bbox)
        
        # Convert normalized bbox to absolute pixel coordinates
        x_min_norm, y_min_norm, w_norm, h_norm = bbox
        x1 = int(x_min_norm * width)
        y1 = int(y_min_norm * height)
        x2 = int((x_min_norm + w_norm) * width)
        y2 = int((y_min_norm + h_norm) * height)

        # Draw bounding box
        display_image = cv2.cvtColor(image.copy(), cv2.COLOR_BGR2RGB)
        cv2.rectangle(display_image, (x1, y1), (x2, y2), (0, 255, 0), 3)
        
        # Display image
        plt.figure(figsize=(10, 10))
        plt.imshow(display_image)
        plt.title(f"Image ID: {id} with Bounding Box")
        plt.axis('off')
        plt.show()
        
    
# Example usage
show_image_with_bbox(100)
```
